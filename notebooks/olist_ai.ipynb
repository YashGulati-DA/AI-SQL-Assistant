````python
def validate_sql(sql):
    sql = sql.strip()

    # Must start with SELECT
    if not sql.upper().startswith("SELECT"):
        return False

    # Remove one optional semicolon at the end
    if sql.endswith(";"):
        sql = sql[:-1].strip()

    # No semicolons should remain
    if ";" in sql:
        return False

    return True
````

````python
print(validate_sql("SELECT * FROM customers"))
````

````python
print(validate_sql("SELECT * FROM customers"))
print(validate_sql("DELETE FROM customers"))
print(validate_sql("SELECT * FROM customers; DELETE FROM customers"))
````

````python
def execute_sql(sql):
    if not validate_sql(sql):
        return "SQL rejected for safety."

    try:
        cursor.execute(sql)
        result = cursor.fetchall()
        return result

    except Exception as e:
        return f"SQL Error: {e}"
````

````python
def format_result(user_question, result):

    if not result:
        return "No results found."

    # Single value
    if len(result[0]) == 1:
        value = result[0][0]

        if isinstance(value, (int, float)):
            return f"The answer is {value:,}."

        return str(value)

    # Multiple rows / multiple columns
    lines = []

    for row in result:
        formatted_row = " | ".join(
            f"{value:,}" if isinstance(value, (int, float)) else str(value)
            for value in row
        )

        lines.append(formatted_row)

    return "\n".join(lines)
````

````python
database_rules = """
Important SQL rules:

1. Use only tables and columns provided in the database schema.
2. Do not invent columns.
3. Do not join tables unless the question requires data from multiple tables.
4. order_items.price contains the item price.
5. order_items.order_item_id is an item sequence number, NOT a quantity or price.
6. For total revenue from order items, use SUM(order_items.price).
7. Generate the simplest correct SQL query possible.
"""
````

````python
def ask_database(user_question):

    # 1. Create SQL prompt
    prompt = f"""
You are an SQL assistant.

Database information:
{database_info}

{database_rules}

User question:
{user_question}

Generate only the SQL query needed to answer the question.
Do not explain anything.
"""

    # 2. Ask Llama to generate SQL
    response = ollama.chat(
        model="llama3.2:3b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    sql = response["message"]["content"].strip()

    # 3. Remove markdown code fences
    if sql.startswith("```"):
        sql = sql.replace("```sql", "").replace("```", "").strip()

    # 4. Validate SQL
    if not validate_sql(sql):
        return "The generated SQL was rejected for safety."

    # 5. Execute SQL
    try:
        cursor.execute(sql)
        result = cursor.fetchall()

    except Exception as e:
        return f"I couldn't execute the generated SQL: {e}"

    # 6. Format the exact result
    answer = format_result(user_question, result)

    return answer
````

````python
def fix_sql(user_question, bad_sql, error_message):

    prompt = f"""
You are an SQL debugging assistant.

DATABASE SCHEMA:
{database_info}

IMPORTANT RULES:
{database_rules}

USER QUESTION:
{user_question}

INCORRECT SQL:
{bad_sql}

DATABASE ERROR:
{error_message}

Your job is to completely rewrite the incorrect SQL.

IMPORTANT:
- Do NOT reuse an incorrect table alias just because it appeared in the old query.
- Check every column against the database schema.
- The `products` table DOES NOT contain a `price` column.
- The `order_items` table DOES contain a `price` column.
- For total revenue, use `order_items.price`.
- Do NOT join the products table unless the question requires product information.
- Return ONLY the corrected SQL query.
"""

    response = ollama.chat(
        model="llama3.2:3b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    sql = response["message"]["content"].strip()

    if sql.startswith("```"):
        sql = sql.replace("```sql", "").replace("```", "").strip()

    return sql
````

````python
bad_sql = """
SELECT SUM(p.price)
FROM order_items o
JOIN products p
ON o.product_id = p.product_id
"""

error_message = "no such column: p.price"

corrected_sql = fix_sql(
    "What is the total revenue from all orders?",
    bad_sql,
    error_message
)

print(corrected_sql)
````

````python
corrected_result = execute_sql(corrected_sql)

print(corrected_result)
````

````python
def ask_database(user_question):

    # 1. Generate SQL
    sql = generate_sql(user_question)

    # 2. Validate SQL
    if not validate_sql(sql):
        return "The generated SQL was rejected for safety."

    # 3. Try executing the SQL
    try:
        cursor.execute(sql)
        result = cursor.fetchall()

    except Exception as e:

        # 4. If SQL fails, ask Llama to fix it
        error_message = str(e)

        corrected_sql = fix_sql(
            user_question,
            sql,
            error_message
        )

        # 5. Validate corrected SQL
        if not validate_sql(corrected_sql):
            return "The corrected SQL was rejected for safety."

        # 6. Try corrected SQL
        try:
            cursor.execute(corrected_sql)
            result = cursor.fetchall()

            sql = corrected_sql

        except Exception as e2:
            return f"SQL failed after correction: {e2}"

    # 7. Format exact database result
    answer = format_result(
        user_question,
        result
    )

    return answer
````
